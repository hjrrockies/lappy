```python
from lappy import *
import numpy as np
import matplotlib.pyplot as plt
```

# Isosceles triangle

```python
phi = np.pi/5
h = np.tan(phi)
iso_tri = geometry.iso_tri(h)

iso_tri.plot()
plt.show()
```

```python
np.rad2deg(iso_tri.int_angles)
```

```python
tri_data = np.load("../results/fb_scoring_iso_tri.npz")
```

```python
tri_data['N'].shape, tri_data['M'].shape
```

```python
fig, axs = plt.subplots(3,5, figsize=(15,9), sharex=True, sharey=True)
axs = axs.flatten()
for i in range(len(tri_data['N'])):
    axs[i].loglog(tri_data['M'], tri_data['sigmas'][i], label='tensions')
    axs[i].legend(fontsize=10)
    axs[i].set_xlabel(r'$m_{\text{singular}}$', fontsize=15)
    axs[i].set_title(f"# basis functions: {tri_data['N'][i]}", fontsize=15)
plt.tight_layout(rect=[0, 0, 1, .95])
fig.suptitle("Isosceles triangle: singular corner weight vs tensions", fontsize=20)
plt.show()
```
